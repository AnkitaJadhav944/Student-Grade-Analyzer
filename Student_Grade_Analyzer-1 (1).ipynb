# Student Grade Analyzer Using Python

**Week 1 project**

This project analyzes marks in five subjects, calculates total and average marks, assigns a grade, and determines PASS/FAIL status.

Run each cell from top to bottom.

## Project code
The code uses variables, data types, conditionals, loops, functions, lists, dictionaries, and exception handling.

```python
# Student Grade Analyzer Using Python

def calculate_grade(average):
    """Return a grade based on the student's average marks."""
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


def analyze_student(name, marks):
    """Calculate and display a student's academic performance."""
    total = sum(marks.values())
    average = total / len(marks)
    grade = calculate_grade(average)

    # A student must score at least 35 in every subject to pass.
    result = "PASS" if all(mark >= 35 for mark in marks.values()) else "FAIL"

    print("\n========== STUDENT REPORT ==========")
    print("Student Name:", name)
    print("\nSubject-wise Marks:")
    for subject, mark in marks.items():
        print(f"{subject}: {mark:g}")

    print("------------------------------------")
    print(f"Total Marks: {total:g} / {len(marks) * 100}")
    print(f"Average Marks: {average:.2f}%")
    print("Grade:", grade)
    print("Result:", result)
    print("====================================")

    return {
        "name": name,
        "total": total,
        "average": average,
        "grade": grade,
        "result": result
    }


def get_valid_mark(subject):
    """Ask for a mark until the user enters a number from 0 to 100."""
    while True:
        try:
            mark = float(input(f"Enter marks in {subject} (0-100): "))
            if 0 <= mark <= 100:
                return mark
            print("Please enter a mark between 0 and 100.")
        except ValueError:
            print("Please enter a valid number.")


def main():
    print("===== STUDENT GRADE ANALYZER =====")
    name = input("Enter student name: ").strip()

    while not name:
        print("Name cannot be empty.")
        name = input("Enter student name: ").strip()

    subjects = ["Python", "Mathematics", "English", "Science", "Computer"]
    marks = {}

    for subject in subjects:
        marks[subject] = get_valid_mark(subject)

    analyze_student(name, marks)


if __name__ == "__main__":
    main()
```

## Run the application
Run the next cell. Enter a student name and marks from 0 to 100 when prompted.

```python
main()
```
